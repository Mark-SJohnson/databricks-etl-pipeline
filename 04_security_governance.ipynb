```sql
CREATE OR REPLACE FUNCTION mask_fare(fare_amount DOUBLE)
RETURNS DOUBLE
RETURN CASE 
    WHEN is_account_group_member('finance_analyst') THEN fare_amount
    ELSE round(fare_amount, -1)  -- round to nearest 10, to hide exact figures
  END;
```

```sql
ALTER TABLE gold_daily_zip_summary
ALTER COLUMN avg_fare SET Mask MAsk_fare;
```

```sql
CREATE OR REPLACE FUNCTION zip_row_filter(pickup_zip INT)
RETURNS BOOLEAN
RETURN is_account_group_member('northeast_team') OR is_account_group_member('admins');


ALTER TABLE gold_daily_zip_summary
SET ROW FILTER zip_row_filter ON (pickup_zip);
```
